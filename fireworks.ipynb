```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import random

# 1. 畫面設定
fig, ax = plt.subplots(figsize=(8, 5))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.set_xlim(0, 10)
ax.set_ylim(0, 7)
ax.axis("off")

colors = ["red", "orange", "yellow", "cyan", "magenta", "white"]
particles = []
points = ax.scatter([], [], s=[], c=[])

# 2. 建立一場煙火
def create_firework():
    x = random.uniform(1, 9)
    y = random.uniform(3, 6)
    color = random.choice(colors)

    for angle in np.linspace(0, 2 * np.pi, 45, endpoint=False):
        speed = random.uniform(0.04, 0.10)
        particles.append({
            "x": x,
            "y": y,
            "vx": np.cos(angle) * speed,
            "vy": np.sin(angle) * speed,
            "color": color,
            "age": 0
        })

# 3. 每一幀更新粒子
def update(frame):
    if frame % 20 == 0:
        create_firework()

    xs, ys, cs, sizes = [], [], [], []
    alive_particles = []

    for p in particles:
        p["age"] += 1
        p["x"] += p["vx"]
        p["y"] += p["vy"]
        p["vy"] -= 0.002  # 重力：逐漸下墜

        if p["age"] < 55 and p["y"] > 0:
            xs.append(p["x"])
            ys.append(p["y"])
            cs.append(p["color"])
            sizes.append(max(1, 15 * (1 - p["age"] / 55)))
            alive_particles.append(p)

    particles[:] = alive_particles

    points.set_offsets(
        np.column_stack((xs, ys)) if xs else np.empty((0, 2))
    )
    points.set_color(cs)
    points.set_sizes(sizes)
    return points,

# 4. 播放動畫
animation = FuncAnimation(
    fig, update, frames=150,
    interval=40, blit=False, repeat=True
)

plt.close(fig)
display(HTML(animation.to_jshtml()))

```
