```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data =({
    "商品": ["煎包", "饅頭", "豆漿", "蛋餅", "奶茶"],
    "數量": [10, 20, 15, 8, 12],
    "單價": [30, 15, 25, 30, 15]
})

df = pd.DataFrame(data)
df["營收"] = df["數量"] * df["單價"]
print(df)
print("=====================")

x = np.array(["煎包", "饅頭", "豆漿", "蛋餅", "奶茶"])
y = np.array(df["營收"])

plt.bar(x, y)
plt.show()

print("GitHub 第三次修改測試 新增bar條形圖")
print("目前正在練習 Pandas 與 Matplotlib")

```
